```python
import datasets
from datasets import load_dataset
```

```python
dataset = load_dataset("bdstar/twitter-sentiment-analysis")
```

```python
dataset["test"][0]
```

```python
from transformers.pipelines.pt_utils import KeyDataset
from tqdm import tqdm

labels = []
cnt = 0
for output in tqdm(ppl(KeyDataset(dataset["test"], "text"))):
    labels.append(output)
    cnt += 1
    if cnt > 15000:
        break
```

```python
true_labels = [elem["label"] for elem in dataset["test"]]
```

```python
true_labels_ = true_labels[:15000]
```

```python
true_labels_[0]
```

```python
idxs_scores = []
for i, elem in enumerate(list(zip(true_labels_, labels))):
    if elem[1]["label"].lower() != elem[0]:
        idxs_scores.append((i, elem[1]['score']))
```

```python
idxs_scores
```

```python
confusing_samples = [dataset["test"][i] for i, score in idxs_scores if score <= 0.7]
```
